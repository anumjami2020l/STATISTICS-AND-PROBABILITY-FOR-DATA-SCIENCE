# Z-TEST

### Formulate the hypotheses:

* Null hypothesis (Ho): The sample mean age is equal to a known population mean age (μ0).
* Alternative hypothesis (H1): The sample mean age is not equal to the known population mean age.

### Hypotheses:

* Ho: μ=60
* H1: μ≠60

```python
from scipy import stats
import numpy as np

# Known population mean
population_mean = 60

# Calculate sample mean and standard deviation for the 'age' variable
sample_mean = heart_data['age'].mean()
sample_std = heart_data['age'].std(ddof=1)  # Sample standard deviation
sample_size = heart_data['age'].count()

# Perform the Z-test
z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(sample_size))
p_value_z_test = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))

print("Sample mean:", sample_mean)
print("Sample standard deviation:", sample_std)
print("Sample size:", sample_size)
print("Z-test statistic:", z_statistic)
print("p-value:", p_value_z_test)
```

### 1- Set the significance level (α):

    Typically, α=0.05α=0.05.

### 2- Compare p-value to α:

    p-value = 0.2238
    α=0.05α=0.05

Since 0.2238>0.05, we do not reject the null hypothesis.

### 3- Conclusion:

    There is not enough evidence to conclude that the mean age of patients in the sample is significantly different from the known population mean of 60 years.

## Summary:

* Null Hypothesis (Ho): The sample mean age is equal to the population mean age of 60 years.
* Alternative Hypothesis (H1): The sample mean age is not equal to the population mean age of 60 years.
* Significance Level (α): 0.05
* Z-test statistic: 1.2165
* p-value: 0.2238
* Decision: Do not reject the null hypothesis.
* Conclusion: The sample mean age is not significantly different from the population mean age of 60 years

# T-Test

### Formulate the hypotheses:

* Null hypothesis (Ho): The mean age of male patients (μ1) is equal to the mean age of female patients (μ2).
* Alternative hypothesis (H1): The mean age of male patients is not equal to the mean age of female patients.

### Hypotheses:

* Ho: μ1=μ2 (The mean age of male patients is equal to the mean age of female patients)
* Ha: μ1≠μ2 (The mean age of male patients is not equal to the mean age of female patients)

```python
# Separate the data into two groups based on sex
male_age = heart_data[heart_data['sex'] == 1]['age']
female_age = heart_data[heart_data['sex'] == 0]['age']

# Calculate sample means and standard deviations
mean_male = np.mean(male_age)
mean_female = np.mean(female_age)
std_male = np.std(male_age, ddof=1)
std_female = np.std(female_age, ddof=1)

# Perform the independent samples t-test
t_stat, p_value_t_test = stats.ttest_ind(male_age, female_age)

print("Mean age of male patients:", mean_male)
print("Mean age of female patients:", mean_female)
print("Standard deviation of age of male patients:", std_male)
print("Standard deviation of age of female patients:", std_female)
print("t-test statistic:", t_stat)
print("p-value:", p_value_t_test)
```

### 1- Set the significance level (α):

    Typically, α=0.05α=0.05.

### 2- Compare p-value to α:

    p-value = 0.2581
    α=0.05α=0.05

Since 0.2581>0.050.2581>0.05, we do not reject the null hypothesis.

### 3- Conclusion:

    There is not enough evidence to conclude that the mean age of male patients is significantly different from the mean age of female patients.

## Summary:

* Null Hypothesis (Ho): he mean age of male patients is equal to the mean age of female patients.
* Alternative Hypothesis (H1): The mean age of male patients is not equal to the mean age of female patients.
* Significance Level (α): 0.05
* t-test statistic: 1.1330
* p-value: 0.2581
* Decision: Do not reject the null hypothesis.
* Conclusion: The mean age of male patients is not significantly different from the mean age of female patients.

# CHI-SQUARE TEST

### Formulate the hypotheses:

* Null hypothesis (H0): The distribution of death events is equal between males and females from Heart Failure.
* Alternative hypothesis (H1): The distribution of death events is not equal between males and females from Heart Failure.



```python
# Filter data for males
male_data = heart_data[heart_data['sex'] == 1]

# Divide male data based on death event
male_death = male_data[male_data['DEATH_EVENT'] == 1]
male_not_death = male_data[male_data['DEATH_EVENT'] == 0]

# Calculate counts for the contingency table
male_death_count = len(male_death)
male_not_death_count = len(male_not_death)

# Print the counts
print("Male death count:", male_death_count)
print("Male not death count:", male_not_death_count)
```

```python
# Filter data for females
female_data = heart_data[heart_data['sex'] == 0]

# Divide female data based on death event
female_death = female_data[female_data['DEATH_EVENT'] == 1]
female_not_death = female_data[female_data['DEATH_EVENT'] == 0]

# Calculate counts for the contingency table
female_death_count = len(female_death)
female_not_death_count = len(female_not_death)

# Print the counts
print("Female death count:", female_death_count)
print("Female not death count:", female_not_death_count)

```

```python


table = pd.DataFrame({
    'Death Event Observed': [male_death_count, female_death_count],
    'Death Event Expected': [48, 48]
}, index=['Male', 'Female'])

print("\nTable:")
print(table)
```

```python
observed = table['Death Event Observed'].values
expected = table['Death Event Expected'].values

import scipy.stats as stats
from scipy.stats import chisquare
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

# Print results
print(f"\nChi-square statistic: {chi2_stat:.5f}")
print(f"p-value: {p_value:.5f}")

# Degrees of freedom
dof = len(observed) - 1
print(f"Degrees of freedom: {dof}")
```
